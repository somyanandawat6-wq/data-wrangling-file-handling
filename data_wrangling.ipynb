**Data Wrangling & File Handling in Various Formats**

In real-world data analysis, information comes in different file formats like text (.txt TSV (.tsv), and JSON (.json).

Python provides built-in and library-based tools to read, write, and process these formats efficiently.

This helps us learn how to load data from different sources and prepare for analysis and modeling.

## Introduction to File Formats

**Text Files ( `.txt` ):**

- Plain text without structured formatting.
- Used for simple data storage (e.g., logs, notes).
- Handled via Python's built-in `open()` function.

```python
with open('example_test1.txt','w') as file:
    for i in range(1,6):
        file.write(f"This is my {i} sample file\n")
print("File Created Successfully")
```

# Reading a text file using built-in open() function

```python
with open('example_test1.txt','r') as file:
    data = file.read()
    print(data)
```

## Key Concepts:

- `read()`: Reads the entire file content.
- `readlines()`: Returns a list of lines (useful for multi-line files).


# Write in Multiple Lines

```python
lines = [
    "This is my First line",
    "This is my Second line",
    "This is my Third line",
    "This is my Fourth line",
    "This is my Fifth line",
]
with open("example_multi.txt","w") as file:
    for i in lines:
        file.write(i + "\n")
print("Completed Successfully")
```

# A python program to read txt file from specific location

```python
with open("example_multi.txt","r") as file:
    lines = file.readlines()
    for i in lines:
        print(i.strip())
```

## Key Concepts- 
### Absolute/relative paths for file access.
- `strip()`: Removes leading/trailing whitespace (including `\n`)


**TSV (Tab-Separated Values):**

- Stores tabular data with tabs ( `\t` ) as delimiters.
- Similar to CSV but uses tabs instead of commas.
- Ideal for datasets with fields containing commas.


```python
import pandas as pd

data = {
    'Emp_ID': [101, 102, 103, 104],
    'Name': ['Rahul Sharma', 'Priya Patel', 'Amit Singh', 'Neeta Rao'],
    'Department': ['HR', 'Finance', 'IT', 'Marketing'],
    'Salary (₹)': [50000, 65000, 80000, 55000],
    'Joining_Date': ['2020-01-15', '2019-05-22', '2021-11-10', '2022-03-05'],
    'City': ['Mumbai', 'Delhi', 'Bangalore', 'Hyderabad']
}

data
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
df
```

```python
df.to_csv('tab_example.tsv' , sep = '\t', index = False)
```

## Key Concepts:

- `sep='\t'`: Specifies tab as the delimiter.
- `index=False`: Avoids writing row indices (by defaulat).
- `index=True`: If your DataFrame has a custom index (e.g. dates, employee IDs) that you want to preserve.


# Reading the TSV file

```python
df_read = pd.read_csv('tab_example.tsv' , sep = '\t')
print(df_read.head())
```

# Same Data, Different Formats

| Data Example | CSV Format (Commas) | TSV Format (Tabs) |
|---|---|---|
| **Simple Data** | `Apple,1.99,Red` | `Apple    1.99    Red` |
| **Data with Commas** | `"Apple, Green",1.99,Green` | `Apple, Green    1.99    Green` |
| **Data with Quotes** | `"Apple \"Honey\"",2.50,Yellow` | `Apple "Honey"    2.50    Yellow` |
| **Readability** | Harder (quotes/escapes) | Cleaner (no conflicts) |

# When to Use Which?

- **Use CSV when:**
  - Working with Excel/Google Sheets (default format).
  - Data is simple (no commas in text).

- **Use TSV when:**
  - Data contains commas, quotes, or multi-line text.
  - Processing logs, NLP data, or large datasets.

# JSON (JavaScript Object Notation):

- Lightweight data interchange format.
- Uses key-value pairs and nested structures.
- Commonly used for APIs and configuration files.

# Creating a json with multiple lines

```python
import json

data = {
    "employees": [
        {"name": "John", "age": 30, "department": "Sales"},
        {"name": "Meera", "age": 25, "department": "Marketing"},
        {"name": "Rahul", "age": 35, "department": "HR"}
    ]
}

# Writing the dictionary to a JSON file
with open('json_example.json', 'w') as file:
    json.dump(data, file, indent=4)
```

- **`json.dump(data, file, indent=4)`**: This function writes the dictionary `data` into the file in JSON format.

  - **`data`**: This is the dictionary you want to write to the JSON file.

  - **`file`**: This refers to the file object created in the `with` statement. It's where the JSON data will be written.

  - **`indent=4`**: This argument makes the JSON file more readable by adding an indentation of 4 spaces to each level in the data hierarchy. This makes it easier to read when you open the file.

 # Key Concepts:

- **`json.dump()`**: Serializes Python objects to JSON.

- **`indent`**: Improves readability by formatting with spaces.

| Format | Structure | Best For | Python Tools |

| **TXT** | Raw text | Logs, notes | `open()` |
